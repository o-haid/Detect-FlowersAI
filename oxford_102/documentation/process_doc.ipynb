```python
import os
import tarfile
import urllib.request
from scipy.io import loadmat
```

```python
data_dir = '/teamspace/studios/this_studio/oxford_102/documentation'

urls = {
    "segmentations": "https://www.robots.ox.ac.uk/~vgg/data/flowers/102/102segmentations.tgz",
    "imagelabels": "https://www.robots.ox.ac.uk/~vgg/data/flowers/102/imagelabels.mat",
    "data split": "https://www.robots.ox.ac.uk/~vgg/data/flowers/102/setid.mat"
}
```

```python
for name, url in urls.items():
    filename = os.path.basename(url)
    filepath = os.path.join(data_dir, filename)
    urllib.request.urlretrieve(url, filepath)
```

```python
tgz_path = os.path.join(data_dir, "102segmentations.tgz")

with tarfile.open(tgz_path, "r:gz") as tar:
        tar.extractall(path=data_dir)
```

```python
imagelabels_path = os.path.join(data_dir, "imagelabels.mat")
setid_path = os.path.join(data_dir, "setid.mat")

labels_data = loadmat(imagelabels_path)
setid_data = loadmat(setid_path)
```
